```python
import yaml
```

```python
def get_class_label(class_id):
    return f"{class_id+1:06b}"[::-1]
```

```python
def label_to_class_id(label: str):
    return int(label, 2) - 1
```

```python
with open("map-characters.yaml", "w") as f:
    for i in range(0, 63):
        binary = get_class_label(i)
        f.write(f"{i}: '{binary}'\n")
```

```python
label_to_class_id("000001")
```

```python
with open("../configs/nepali.yaml", "r") as f:
    braille = yaml.safe_load(f)

braille
```

```python
labels = {}
```

```python
for k, v in braille.items():
    print(k,v)
    class_id = label_to_class_id(k)

    labels[class_id] = v
```
