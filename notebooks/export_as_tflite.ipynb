```python
tflite_model = YOLO("../weights/deployment/yolo26n/finetuned/best_finetuned_mAP_95_9_saved_model/best_finetuned_mAP_95_9_float16.tflite", task="detect")
```

```python
results = tflite_model.predict("/var/home/supremekhadka/Downloads/braille_images/dear_ebu.jpeg")
```

```python
# Force-extract the first element from the returned results list
result = results[0]  

# Now you can run the vectorized logic without errors
boxes = result.boxes.xyxy.cpu().numpy()     
confs = result.boxes.conf.cpu().numpy()     
clss  = result.boxes.cls.cpu().numpy().astype(int) 

for coords, conf, class_id in zip(boxes, confs, clss):
    x1, y1, x2, y2 = coords
    class_name = result.names[class_id]
    print(f"Detected {class_name} with {conf:.2f} at [{x1}, {y1}]")
```
