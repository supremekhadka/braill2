```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
import statistics
```

## Visualizations on Processed Dataset

```python
character_label_dir = os.listdir("../data/dataset/characters/labels")

all_annots = ""

for split in character_label_dir:
    for label_file in os.listdir(f"../data/dataset/characters/labels/{split}"):
        print(label_file)
        with open(f"../data/dataset/characters/labels/{split}/{label_file}", "r") as f:
            all_annots += f.read()

print(f"Total lines: {len(all_annots)}")
```

```python
parsed_annots = [list(map(float, a.split())) for a in all_annots.split("\n")]

print(parsed_annots[:5])
```

```python
classes = [int(a[0]) for a in parsed_annots if len(a) > 0]
widths = [a[3] for a in parsed_annots if len(a) > 0]
heights = [a[4] for a in parsed_annots if len(a) > 0]

print(max(classes))
print(max(widths))
print(max(heights))

print(len(classes))
print(len(widths))
print(len(heights))
```

```python
mean_width = round(statistics.mean(widths),10)
stdev_width = round(statistics.pstdev(widths), 10)

print(f"Normalized Width: {mean_width} ± {stdev_width}")
```

```python
mean_height = round(statistics.mean(heights),10)
stdev_height = round(statistics.pstdev(heights), 10)

print(f"Normalized Height: {mean_height} ± {stdev_height}")
```

```python
plt.figure(figsize=(15,6))

plt.hist(widths, bins=30, edgecolor='black')

ticks = np.linspace(min(widths), max(widths), 30) 

plt.xlabel("Width")
plt.xticks(ticks, rotation=45, fontsize=9)
plt.ylabel("Frequency")
plt.title("Width Distribution")

plt.savefig("../visualizations/class_distributions/angelina+dsbi_widths_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(15,6))

plt.hist(heights, bins=40, edgecolor='black')

ticks = np.linspace(min(heights), max(heights), 40) 

plt.xlabel("Height")
plt.xticks(ticks, rotation=45, fontsize=9)
plt.ylabel("Frequency")
plt.title("Height Distribution")

plt.savefig("../visualizations/class_distributions/angelina+dsbi_heights_distribution.png")
plt.show()
```
